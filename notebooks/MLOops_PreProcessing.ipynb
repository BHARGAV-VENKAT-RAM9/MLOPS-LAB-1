```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = pd.read_csv(
    "cse_curriculum_industry_skill_alignment_1500.csv"
)

print("Dataset shape:", data.shape)

data.head()
```

```python
print(data.columns.tolist())
```

```python
data.info()
```

```python
print(data.isnull().sum())
```

```python
target = "placement_status"
```

```python
features = [
    "university",
    "gender",
    "graduation_year",
    "target_role",
    "cgpa",
    "attendance_percent",
    "internship_count",
    "project_count",
    "certification_count",
    "hackathon_count",
    "programming_language",
    "curriculum_programming_score",
    "curriculum_dsa_score",
    "curriculum_dbms_score",
    "curriculum_web_development_score",
    "curriculum_cloud_score",
    "curriculum_ai_ml_score",
    "curriculum_cybersecurity_score",
    "curriculum_devops_score",
    "curriculum_communication_score",
    "curriculum_problem_solving_score"
]

X = data[features]
y = data[target]

print("Features:")
display(X.head())

print("Target:")
display(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
```

```python
numerical_only = [
    "graduation_year",
    "cgpa",
    "attendance_percent",
    "internship_count",
    "project_count",
    "certification_count",
    "hackathon_count",
    "curriculum_programming_score",
    "curriculum_dsa_score",
    "curriculum_dbms_score",
    "curriculum_web_development_score",
    "curriculum_cloud_score",
    "curriculum_ai_ml_score",
    "curriculum_cybersecurity_score",
    "curriculum_devops_score",
    "curriculum_communication_score",
    "curriculum_problem_solving_score"
]
```

```python
baseline_train = X_train[numerical_only].copy()
baseline_test = X_test[numerical_only].copy()

train_valid_rows = baseline_train.dropna().index
test_valid_rows = baseline_test.dropna().index

baseline_train = baseline_train.loc[train_valid_rows]
baseline_test = baseline_test.loc[test_valid_rows]

y_train_baseline = y_train.loc[train_valid_rows]
y_test_baseline = y_test.loc[test_valid_rows]
```

```python
baseline_model = LogisticRegression(
    max_iter=2000
)

baseline_model.fit(
    baseline_train,
    y_train_baseline
)
```

```python
baseline_predictions = baseline_model.predict(
    baseline_test
)
```

```python
baseline_accuracy = accuracy_score(
    y_test_baseline,
    baseline_predictions
)

print(
    "Baseline accuracy:",
    round(baseline_accuracy * 100, 2),
    "%"
)

print(
    "Training records used:",
    len(baseline_train)
)

print(
    "Testing records used:",
    len(baseline_test)
)
```

```python
numerical_features = [
    "graduation_year",
    "cgpa",
    "attendance_percent",
    "internship_count",
    "project_count",
    "certification_count",
    "hackathon_count",
    "curriculum_programming_score",
    "curriculum_dsa_score",
    "curriculum_dbms_score",
    "curriculum_web_development_score",
    "curriculum_cloud_score",
    "curriculum_ai_ml_score",
    "curriculum_cybersecurity_score",
    "curriculum_devops_score",
    "curriculum_communication_score",
    "curriculum_problem_solving_score"
]
```

```python
categorical_features = [
    "university",
    "gender",
    "target_role",
    "programming_language"
]

print("Categorical features defined:", len(categorical_features))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numerical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

print("Numerical pipeline created successfully.")
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

print("Categorical pipeline created successfully.")
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_pipeline,
            numerical_features
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features
        )
    ]
)

print("Preprocessor created successfully.")
```

```python
complete_pipeline = Pipeline(
    steps=[
        (
            "preprocessing",
            preprocessor
        ),
        (
            "model",
            LogisticRegression(
                max_iter=2000
            )
        )
    ]
)
```

```python
complete_pipeline.fit(
    X_train,
    y_train
)

print("Pipeline training completed successfully.")
```

```python
processed_predictions = complete_pipeline.predict(
    X_test
)
```

```python
processed_accuracy = accuracy_score(
    y_test,
    processed_predictions
)

print(
    "Accuracy after preprocessing:",
    round(processed_accuracy * 100, 2),
    "%"
)
```

```python
print(
    classification_report(
        y_test,
        processed_predictions
    )
)
```

```python
comparison = pd.DataFrame(
    {
        "Approach": [
            "Before proper preprocessing",
            "After preprocessing pipeline"
        ],
        "Accuracy": [
            baseline_accuracy,
            processed_accuracy
        ],
        "Training records used": [
            len(baseline_train),
            len(X_train)
        ],
        "Testing records used": [
            len(baseline_test),
            len(X_test)
        ]
    }
)

display(comparison)
```

```python
X_train_processed = complete_pipeline.named_steps[
    "preprocessing"
].transform(X_train)

X_test_processed = complete_pipeline.named_steps[
    "preprocessing"
].transform(X_test)
```

```python
feature_names = complete_pipeline.named_steps[
    "preprocessing"
].get_feature_names_out()

print("Number of processed features:", len(feature_names))

print(feature_names)
```

```python
processed_train_df = pd.DataFrame(
    X_train_processed,
    columns=feature_names,
    index=X_train.index
)

processed_test_df = pd.DataFrame(
    X_test_processed,
    columns=feature_names,
    index=X_test.index
)
```

```python
processed_train_df["placement_status"] = y_train.values
processed_test_df["placement_status"] = y_test.values
```

```python
display(processed_train_df.head())
```

```python
processed_train_df.to_csv(
    "cse_processed_train.csv",
    index=False
)

processed_test_df.to_csv(
    "cse_processed_test.csv",
    index=False
)

print("Processed datasets saved successfully.")
```
